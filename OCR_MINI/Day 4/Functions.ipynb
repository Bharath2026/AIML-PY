**Local and Global Variable**

```python
def myfunc1():
    x = "Jane"
    def myfunc2():
        x = "hello"
        print(f"Value of x inside myfunc2: {x}")
    return myfunc2
word = myfunc1()
word()
```

```python
class Person():
  def person1(self, name, age):
    self.name = name
    self.age = age
P1=Person()
P1.name = "Bharath"
P1.age = 23
print(P1.name)
print(P1.age)
del P1.age
print(P1.age)
```
