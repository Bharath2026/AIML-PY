```python
import numpy as np

matrix = np.random.randint(1, 51, size=(5, 5))
print("Original matrix:")
print(matrix)

print("Numbers greater than 25:")
print(matrix[matrix > 25])

new_matrix = matrix.copy()
new_matrix[new_matrix % 2 == 0] = 0
print("Modified matrix:")
print(new_matrix)

arr = np.random.randint(1, 101, size=10)
minimum = np.min(arr)
maximum = np.max(arr)

if maximum != minimum:
    normalized = (arr - minimum) / (maximum - minimum)
else:
    normalized = np.zeros_like(arr, dtype=float)

print("Original 1D array:", arr)
print("Normalized array:", normalized)
```

```python
import pandas as pd
import numpy as np
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Ravi", "Priya", "Amit", "Sneha", "Kiran", "Anu"],
    "Department": ["IT", "HR", "IT", "HR", "Finance", "Finance"],
    "Age": [25, 28, 30, 26, 32, 29],
    "Salary": [50000, 60000, np.nan, np.nan, 55000, 65000]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("Missing values in each column:")
print(df.isnull().sum())

df["Salary"] = df["Salary"].fillna(
    df.groupby("Department")["Salary"].transform("median")
)
print(df)
df["Annual_Bonus"] = df["Salary"] * 0.10
print("DataFrame with Annual Bonus:")
print(df)

average_salary = df.groupby("Department")["Salary"].mean()

print("Average salary per department:")
print(average_salary)
```

```python
arr = np.arange(1, 21)
print("1D Array:")
print(arr)
matrix = arr.reshape(4, 5)
print("\n4x5 Matrix:")
print(matrix)
columns = matrix[:, 1:3]
print("\n2nd and 3rd columns:")
print(columns)
corners = matrix[[0, 0, 3, 3], [0, 4, 0, 4]]
print("\nFour corner elements:")
print(corners)
```

```python
data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones", "Printer"],
    "Category": ["Electronics", "Accessories", "Accessories",
                 "Electronics", "Accessories", "Electronics"],
    "Price": [50000, 300, 800, 12000, 1500, 7000],
    "Stock": [10, 100, 35, 25, 15, 40]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
filtered_df = df[(df["Price"] > 500) & (df["Stock"] < 50)]

print("Products with Price > 500 and Stock < 50:")
print(filtered_df)
df["Restock_Needed"] = df["Stock"] < 20
print("DataFrame with Restock_Needed:")
print(df)
df = df.sort_values(by="Price", ascending=False)
print("Final DataFrame:")
print(df)
```

```python
celsius = np.random.uniform(15.0, 40.0, 10)
print("Celsius temperatures:")
print(celsius)
fahrenheit = (celsius * 9 / 5) + 32
print("Fahrenheit temperatures:")
print(fahrenheit)
arr = np.arange(1, 11)
square_roots = np.sqrt(arr)
print("Original array:")
print(arr)
print("Square roots:")
print(square_roots)
max_index = np.argmax(fahrenheit)
print("Index of maximum Fahrenheit temperature:")
print(max_index)
print("Maximum Fahrenheit temperature:")
print(fahrenheit[max_index])
```
