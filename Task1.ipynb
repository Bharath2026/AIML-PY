```python
Grades= [78,85,62,90,85,95,78]
Grades_1 = set(Grades)
print(Grades_1)
print(sorted(list(Grades_1), reverse=True))
grades=list(Grades_1)
print(grades)
minimum=100
maximum=0
for grade in grades:
  if grade<minimum:
    minimum=grade
  if grade>maximum:
    maximum=grade
print(minimum,maximum)
print("average marks = ", sum(grades)/len(grades))
```

```python
classroom = {
    "S101": {"name": "Alice", "subjects": {"Math": 88, "Science": 92}},
    "S102": {"name": "Bob", "subjects": {"Math": 75, "Science": 80}}
}

classroom["S103"] = {"name": "Charlie", "subjects": {"Math": 95, "Science": 89}}
alice_scores = classroom["S101"]["subjects"].values()
alice_average = sum(alice_scores) / len(alice_scores)
print("Alice's average score: ",alice_average)

classroom["S102"]["subjects"]["Math"] = 82

print("\nUpdated Classroom Dictionary:")
print(classroom)
```

```python
bill_amount = 125.50
tax_percentage = 8
tip_percentage = 15
tax_amount = bill_amount * (tax_percentage / 100)
tip_amount = bill_amount * (tip_percentage / 100)
grand_total = bill_amount + tax_amount + tip_amount

print("Original Bill: $", round(bill_amount, 2))
print("Tax (", tax_percentage, "%) $", round(tax_amount, 2))
print("Tip (", tip_percentage, "%) $", round(tip_amount, 2))
print("Grand Total: $", round(grand_total, 2))
```

```python
inventory = {
    "laptop": 12,
    "mouse": 50,
    "keyboard": 25,
    "monitor": 8
}
inventory["headphones"] = 30
inventory["mouse"] = 45
del inventory["monitor"]
in_stock_items = set(inventory.keys())
discounted_items = {"mouse", "keyboard", "webcam"}
print("Updated Inventory:", inventory)
print("In-stock Items:", in_stock_items)
print("Discounted Items:", discounted_items)
print("\nItems both in stock and discounted:", in_stock_items.intersection(discounted_items))
print("Discounted items out of stock:", discounted_items.difference(in_stock_items))
```

```python
sales = {
 "Monday": {"apples": 10, "bananas": 15, "oranges": 8},
 "Tuesday": {"apples": 12, "bananas": 20, "oranges": 5},
 "Wednesday": {"apples": 8, "bananas": 18, "oranges": 12}
}

total_apples_sold = 0
total_bananas_sold = 0
for day_sales in sales.values():
    total_apples_sold += day_sales.get("apples", 0)
    total_bananas_sold += day_sales.get("bananas", 0)
print("Total apples sold: ",total_apples_sold)
print("Total bananas sold: ",total_bananas_sold)
if total_apples_sold > total_bananas_sold:
    print("Apples sold more in total.")
elif total_bananas_sold > total_apples_sold:
    print("Bananas sold more in total.")
else:
    print("Apples and bananas sold the same amount in total.")
```
