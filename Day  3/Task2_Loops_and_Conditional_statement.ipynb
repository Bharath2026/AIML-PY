Temperature Alert (if-elif-else)

Problem: Write a program where temperature > 30 prints 'Hot', between 15 and 30 prints 'Warm', and < 15 prints
'Cold'.


```python
Temperature = int(input("Enter Your Temperature:"))
if Temperature > 30:
  print("HOT")
elif Temperature >= 15:
  print("Warm")
else:
  print("Cold")
```

Sum of Even Numbers (for loop)
Problem: Calculate the sum of all even numbers from 1 to 20 using range().

```python
total=0
for n in range(2,21,2):
  total+=n
print(total)

```

Multiplication Table Grid (Nested Loops) Problem:

Print a 5x5 grid displaying multiplication tables for numbers 1 to 5

```python
for i in range(1,6):
  for j in range(1,6):
    print(i*j, end=" ")
  print()
```
