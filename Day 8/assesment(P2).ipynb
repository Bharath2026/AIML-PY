```python
def compress_string(s):
    compressed = ""
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            compressed += s[i - 1] + str(count)
            count = 1
    if s:
        compressed += s[-1] + str(count)
    if len(compressed) >= len(s):
        return s
    return compressed
s = "aabcccccaaa"
print(compress_string(s))
```

```python
def longest_consecutive(nums):
    numbers = set(nums)
    longest = 0
    for num in numbers:
        if num - 1 not in numbers:
            length = 1
            while num + length in numbers:
                length += 1
            longest = max(longest, length)
    return longest

print(longest_consecutive([100, 4, 200, 1, 3, 2]))
```

```python
def moving_average(data, window_size):
    result = []
    for i in range(len(data) - window_size + 1):
        window = data[i:i + window_size]
        average = sum(window) / window_size
        result.append(round(average, 2))
    return result
print(moving_average([1, 2, 3, 4, 5], 3))
```

```python
def generate_pascals_triangle(n):
    triangle = []
    for i in range(n):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]
        triangle.append(row)
    return triangle
print(generate_pascals_triangle(5))
```

```python
def group_anagrams(words):
    groups = {}
    for word in words:
        key = ''.join(sorted(word))
        if key not in groups:
            groups[key] = []
        groups[key].append(word)
    return groups
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(group_anagrams(words))
```

```python
def align_records(records):
    keys = set()
    for record in records:
        keys.update(record.keys())
    result = []
    for record in records:
        new_record = {}
        for key in keys:
            new_record[key] = record.get(key, None)
        result.append(new_record)
    return result
records = [
    {"name": "Alice", "age": 20},
    {"name": "Bob", "salary": 50000}
]
print(align_records(records))
```

```python
class DimensionMismatchError(Exception):
    pass
def matrix_multiply(A, B):
    if len(A[0]) != len(B):
        raise DimensionMismatchError("Invalid matrix dimensions")
    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(B[0])):
            total = 0
            for k in range(len(B)):
                total += A[i][k] * B[k][j]
            row.append(total)
        result.append(row)
    return result
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print(matrix_multiply(A, B))
```

```python
def extract_nested_keys(data, target_key):
    result = []
    if isinstance(data, dict):
        for key, value in data.items():
            if key == target_key:
                result.append(value)
            if isinstance(value, (dict, list)):
                result.extend(extract_nested_keys(value, target_key))
    elif isinstance(data, list):
        for item in data:
            result.extend(extract_nested_keys(item, target_key))
    return result
data = {
    "name": "Alice",
    "details": {
        "name": "Bob",
        "age": 20
    }
}

print(extract_nested_keys(data, "name"))
```

```python
def pivot_summary(data, group_by_col, value_col):
    groups = {}
    for row in data:
        key = row[group_by_col]
        if key not in groups:
            groups[key] = []
        groups[key].append(row[value_col])
    result = {}
    for key, values in groups.items():
        result[key] = {
            "sum": sum(values),
            "average": sum(values) / len(values),
            "count": len(values)
        }
    return result
data = [
    {"department": "IT", "salary": 50000},
    {"department": "IT", "salary": 60000},
    {"department": "HR", "salary": 40000}
]

print(pivot_summary(data, "department", "salary"))
```

```python
def process_tasks(tasks):
    tasks.sort(key=lambda task: task["priority"])
    result = []
    for task in tasks:
        result.append(task["name"])
    return result
tasks = [
    {"name": "Task A", "priority": 3},
    {"name": "Task B", "priority": 1},
    {"name": "Task C", "priority": 2}
]

print(process_tasks(tasks))
```

```python
import time
def time_execution(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Execution time:",
              (end - start) * 1000, "ms")
        return result
    return wrapper
@time_execution
def test():
    time.sleep(1)
test()
```

```python
def enforce_types(type_schema):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for name, value in kwargs.items():
                if name in type_schema:
                    if not isinstance(value, type_schema[name]):
                        raise TypeError(
                            f"{name} must be {type_schema[name]}"
                        )
            return func(*args, **kwargs)
        return wrapper
    return decorator
@enforce_types({"val": float})
def calculate(val):
    return val * 2

print(calculate(val=5.5))
```

```python
def memoize(func):
    cache = {}
    def wrapper(n):
        if n not in cache:
            cache[n] = func(n)
        return cache[n]
    return wrapper
@memoize
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
print(fibonacci(10))
```

```python
class Array2D:
    def __init__(self, data, rows, cols):
        self.data = data
        self.rows = rows
        self.cols = cols
    def __getitem__(self, index):
        row, col = index
        return self.data[row * self.cols + col]
    def __setitem__(self, index, value):
        row, col = index
        self.data[row * self.cols + col] = value
    def reshape(self, new_rows, new_cols):
        if new_rows * new_cols != len(self.data):
            raise ValueError("Invalid dimensions")
        return Array2D(self.data, new_rows, new_cols)
arr = Array2D([1, 2, 3, 4, 5, 6], 2, 3)
print(arr[1, 2])
arr[1, 2] = 10
print(arr[1, 2])
```

```python
from abc import ABC, abstractmethod


class BasePipelineStep(ABC):

    @abstractmethod
    def process(self, data):
        pass


class FillMissingStep(BasePipelineStep):
    def __init__(self, default):
        self.default = default
    def process(self, data):
        return [
            self.default if x is None else x
            for x in data
        ]
class ScaleStep(BasePipelineStep):
    def __init__(self, factor):
        self.factor = factor
    def process(self, data):
        return [x * self.factor for x in data]
class Pipeline:
    def __init__(self, steps):
        self.steps = steps
    def execute(self, data):
        for step in self.steps:
            data = step.process(data)
        return data
pipeline = Pipeline([
    FillMissingStep(0),
    ScaleStep(2)
])
print(pipeline.execute([10, None, 20]))
```

```python
class AccountLedger:
    def __init__(self):
        self.__transactions = []
    def add_transaction(self, amount):
        self.__transactions.append(amount)
    @property
    def balance(self):
        return sum(self.__transactions)
    @property
    def transaction_count(self):
        return len(self.__transactions)
account = AccountLedger()
account.add_transaction(1000)
account.add_transaction(-200)
account.add_transaction(500)
print(account.balance)
print(account.transaction_count)
```

```python
from abc import ABC, abstractmethod
import json
import csv
class DataExporter(ABC):
    @abstractmethod
    def export(self, data, filepath):
        pass
class JSONExporter(DataExporter):
    def export(self, data, filepath):
        with open(filepath, "w") as file:
            json.dump(data, file, indent=4)
class CSVExporter(DataExporter):
    def export(self, data, filepath):
        with open(filepath, "w", newline="") as file:
            writer = csv.DictWriter(file, fieldnames=data[0].keys())
            writer.writeheader()
            writer.writerows(data)
class ExporterFactory:
    @staticmethod
    def get_exporter(format_type):
        if format_type == "json":
            return JSONExporter()
        if format_type == "csv":
            return CSVExporter()
        raise ValueError("Unsupported format")
data = [
    {"name": "Alice", "age": 20},
    {"name": "Bob", "age": 22}
]

exporter = ExporterFactory.get_exporter("json")
exporter.export(data, "data.json")
```

```python
class GroupedData:
    def __init__(self, data, column):
        self.data = data
        self.column = column
    def sum(self, value_column):
        result = {}
        for row in self.data:
            key = row[self.column]
            if key not in result:
                result[key] = 0
            result[key] += row[value_column]
        return result
    def mean(self, value_column):
        groups = {}
        for row in self.data:
            key = row[self.column]
            if key not in groups:
                groups[key] = []
            groups[key].append(row[value_column])
        result = {}
        for key, values in groups.items():
            result[key] = sum(values) / len(values)
        return result
```
