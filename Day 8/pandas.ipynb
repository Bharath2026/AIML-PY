```python
import pandas as pd
dict1 = {
    "name": ["Ford Mustang", "BMW", "BYD", "Supra", "Lamborgini", "Porsche"],
    "seat": [4, 5, 5, 2, 2, 2]
}
dict2=pd.DataFrame(dict1)
print(dict1)
print(dict2)
print("Version:", pd.__version__)

pd.Series(dict1)

print(dict2['name'])

print(pd.Series(dict1,index=[1,2,3,4,5,6]))
```

```python
a=["a","b","c","d"]
mypd=pd.Series(a)
print(mypd)
```

```python
a=["a","b","c","d"]
mypd=pd.Series(a,index=[1,2,3,4])
print(mypd[1])
```
